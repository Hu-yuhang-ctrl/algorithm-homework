## 问题 3

给定一个包含n个元素的数组 `profits` , 它的第 `i` 个元素 `profits[i]` 表示一支股票第 `i` 天的**收益**（正数表示涨, 负数表示跌）。你只能选择 **某一天** 买入这只股票, 并选择在 **未来的某一个不同的日子** 卖出该股票。

1. 设计一个算法来计算你所能获取的最大利润和对应买入和卖出的日期。请分析算法方案, 计算其时间复杂度, 并且使用python编程实现该算法。

2. \* 设计一个时间复杂度为 $O(n)$的算法实现该算法

e.g. :
---
profits=[3,2,1,-7,5,2,-1,3,-1], 第5天买入, 第8天卖出, 收益最大：9

idea:

**题目分析**

&emsp;&emsp;题目给出数组：

$$
profits=[3,2,1,-7,5,2,-1,3,-1]
$$

&emsp;&emsp;其中 `profits[i]` 表示第 $i$ 天股票的收益，正数表示上涨，负数表示下跌。

&emsp;&emsp;现在需要选择一个买入日期和一个未来的卖出日期，使得从买入日到卖出日之间的累计收益最大。

**算法方案**

&emsp;&emsp;本题采用枚举法。

&emsp;&emsp;因为买入日必须早于卖出日，所以本题可以直接枚举所有可能的买入、卖出组合。

&emsp;&emsp;先固定一个买入日，然后尝试它后面的所有卖出日。

&emsp;&emsp;比如先假设第 1 天买入，再分别尝试第 2 天、第 3 天、第 4 天……卖出，计算每种情况下的收益；然后再假设第 2 天买入，再尝试第 3 天、第 4 天、第 5 天……卖出。

&emsp;&emsp;这样把所有可能情况都检查一遍，最后留下收益最大的一组。

**算法核心思想**

&emsp;&emsp;把所有合法的买入、卖出日期组合都尝试一遍，从中选出收益最大的方案。

**时间复杂度分析**

&emsp;&emsp;设数组长度为 $n$。

&emsp;&emsp;外层循环枚举买入日期，内层循环枚举卖出日期。

&emsp;&emsp;对于每一个买入日期，卖出日期必须在买入日期之后，所以需要检查的组合数量大约是：

$$
n+(n-1)+(n-2)+\cdots+1
$$

&emsp;&emsp;这是一个等差数列，求和为：

$$
\frac{n(n-1)}{2}
$$

&emsp;&emsp;展开后是：

$$
\frac{n^2-n}{2}
$$

&emsp;&emsp;在时间复杂度分析中，主要关注增长最快的部分。这里增长最快的是：

$$
n^2
$$

&emsp;&emsp;因此，该算法的时间复杂度为：

$$
O(n^2)
$$

&emsp;&emsp;通俗地说，如果天数增加很多，需要检查的买入和卖出组合数量会明显变多。

&emsp;&emsp;例如：如果有 10 天，可能要检查几十种买卖组合；如果有 100 天，就可能要检查几千种组合；如果有 1000 天，就可能要检查几十万种组合。

&emsp;&emsp;因此，这个算法在数据量较小时比较容易使用，但数据量很大时效率不高。<br>

（本题在解题过程中参考了大模型的做题思路）

````python
# add your code here
def find_max_profit(profits):
    """
    功能：
        在 profits 数组中，找到最大累计收益，以及对应的买入日期和卖出日期。

    参数：
        profits: list[int]
            profits[i] 表示第 i+1 天的股票收益。

    返回：
        max_profit: int    最大累计收益
        best_buy_day: int  最佳买入日期，从 1 开始计数
        best_sell_day: int 最佳卖出日期，从 1 开始计数
    """

    # 第一步：获取数组长度
    n = len(profits)    # n 表示一共有多少天

    # 第二步：判断是否至少有两天
    # 因为题目要求：必须选择某一天买入，并在未来某一个不同的日子卖出，所以至少需要 2 天才能完成一次买卖
    if n < 2:
        raise ValueError("数组长度至少要为 2，才能选择买入日和卖出日。")

    # 第三步：初始化最大收益
    # 这里不能初始化为 0
    # 因为如果所有收益都是负数，最大收益也可能是负数，此时如果初始化为0，程序就会错误的返回0。
    # 所以先设为负无穷，保证后面第一个合法方案一定可以更新它
    max_profit = float("-inf")

    # 第四步：初始化最佳买入日期和卖出日期
    # 这里先设为 None，后面找到更好的方案时再更新
    best_buy_day = None
    best_sell_day = None

    # 第五步：枚举买入日期
    # i 表示买入日对应的 Python 下标
    # 因为卖出日必须在买入日之后，所以买入日最多只能到倒数第二天
    for i in range(n - 1):

        # 第六步：初始化当前累计收益
        # 假设从第 i 天买入，那么当前收益先从 profits[i] 开始
        current_profit = profits[i]

        # 第七步：枚举卖出日期
        # j 表示卖出日对应的 Python 下标
        # 因为卖出日必须晚于买入日，所以 j 从 i+1 开始
        for j in range(i + 1, n):

            # 第八步：把第 j 天的收益加入当前累计收益
            # 这样 current_profit 就表示从第 i 天到第 j 天的总收益
            current_profit += profits[j]

            # 第九步：判断当前方案是否比之前的最大收益更好
            # 如果当前累计收益更大，就更新最大收益
            if current_profit > max_profit:
                max_profit = current_profit

                # 第十步：记录买入日期和卖出日期
                # 因为 Python 下标从 0 开始，而题目日期从 1 开始
                # 所以要加 1
                best_buy_day = i + 1
                best_sell_day = j + 1

    # 第十一步：返回最终结果
    return max_profit, best_buy_day, best_sell_day


# 题目给出的示例数组
profits = [3, 2, 1, -7, 5, 2, -1, 3, -1]

# 调用函数，得到最大收益、最佳买入日期和最佳卖出日期
max_profit, buy_day, sell_day = find_max_profit(profits)

# 输出结果
print("最大收益为：", max_profit)
print("最佳买入日期：第", buy_day, "天")
print("最佳卖出日期：第", sell_day, "天")
````

## 问题 4

给定一个包含不同整数元素的数组 $ A[1..n]$ ,并且满足条件：$A[1]>A[2]$ 并且 $A[n-1]<A[n]$; 规定：如果一个元素比它两边的邻居元素都小, 即：$A[x]<A[x-1], A[x]<A[x+1]$ , 称这个元素A[x]为“局部最小”。通过遍历一次数组, 我们可以很容易在 $O(n)$的时间复杂度下找到一个局部最小值, 


1. 分析该问题, 设计一个算法在$O(logn)$的时间复杂度下找到一个局部最小(返回数值), 要求：分析算法设计思路, 并且使用python编程实现
2. \* 设计算法找出所有局部最小值, 分析算法设计思路, 并使用python编程实现

e.g.:
---
A=[9, 3, 7, 2, 1, 4, 5 ] 时,  局部最小元素为 3, 1

idea:

**算法设计思路**

*1. 设计思路*

&emsp;&emsp;如果直接从左到右遍历整个数组，依次判断每个元素是否小于左右相邻元素，虽然可以找到局部最小值，但是这种方法的时间复杂度是：

$$
O(n)
$$

&emsp;&emsp;而题目要求时间复杂度为：

$$
O(\log n)
$$

&emsp;&emsp;因此，本题不能使用普通遍历的方法，而应该采用类似二分查找的思想。

&emsp;&emsp;***本题的核心思想是***：每次查看数组中间位置 `mid` 的元素，然后比较它与左右相邻元素的大小关系。

&emsp;&emsp;如果满足：

<div align="center">

`A[mid] < A[mid - 1] and A[mid] < A[mid + 1]`

</div>

&emsp;&emsp;说明 `A[mid]` 同时小于左右两个相邻元素，因此 `A[mid]` 就是一个局部最小值，可以直接返回。

&emsp;&emsp;如果满足：

<div align="center">

`A[mid] > A[mid - 1]`

</div>

&emsp;&emsp;说明 `mid` 左边的相邻元素更小，此时可以向左半部分继续查找。

&emsp;&emsp;如果满足：

<div align="center">

`A[mid] > A[mid + 1]`

</div>

&emsp;&emsp;说明 `mid` 右边的相邻元素更小，此时可以向右半部分继续查找。

&emsp;&emsp;这样每次都可以排除一半的搜索区间，因此能够达到 $O(\log n)$ 的时间复杂度。

*2. 为什么局部最小值一定存在？*

&emsp;&emsp;题目已经给出：

$$
A[1] > A[2]
$$

&emsp;&emsp;这说明数组最左边一开始是下降的。

&emsp;&emsp;同时题目又给出：

$$
A[n-1] < A[n]
$$

&emsp;&emsp;这说明数组最右边最后是上升的。

&emsp;&emsp;因此，数组从左侧的下降趋势，到右侧的上升趋势，中间一定会出现某个“谷底”位置。这个谷底元素就会比左右相邻元素都小，所以一定存在局部最小值。

*3. 算法步骤*

&emsp;&emsp;由于局部最小值需要同时有左邻居和右邻居，因此第一个元素和最后一个元素不作为查找对象。

&emsp;&emsp;在 Python 中，数组下标从 `0` 开始，所以查找范围为：

$$
1 \sim n-2
$$

&emsp;&emsp;具体步骤如下：

&emsp;&emsp;第一步，设置左边界：

$$
left = 1
$$

&emsp;&emsp;设置右边界：

$$
right = n - 2
$$

&emsp;&emsp;第二步，取当前查找区间的中间位置：

$$
mid = \frac{left + right}{2}
$$

&emsp;&emsp;在 Python 中写作：

<div align="center">

`mid = (left + right) // 2`

</div>

&emsp;&emsp;第三步，判断 `A[mid]` 是否同时小于左右相邻元素。

&emsp;&emsp;如果满足：

<div align="center">

`A[mid] < A[mid - 1] and A[mid] < A[mid + 1]`

</div>

&emsp;&emsp;则说明 `A[mid]` 是局部最小值，直接返回。

&emsp;&emsp;第四步，如果：

<div align="center">

`A[mid] > A[mid - 1]`

</div>

&emsp;&emsp;说明左边相邻元素更小，所以局部最小值一定可以在左半部分找到，此时更新右边界：

<div align="center">

`right = mid - 1`

</div>

&emsp;&emsp;第五步，否则说明右边相邻元素更小，所以局部最小值一定可以在右半部分找到，此时更新左边界：

<div align="center">

`left = mid + 1`

</div>

&emsp;&emsp;不断重复上述过程，直到找到一个局部最小值。

**算法时间复杂度分析**

&emsp;&emsp;本题使用的是类似二分查找的方法。

&emsp;&emsp;每次查找时，算法都会取当前区间的中间位置 `mid`，然后根据 `A[mid]` 与左右相邻元素的大小关系，决定去左半部分还是右半部分继续查找。

&emsp;&emsp;因此，每经过一次循环，查找范围都会大约缩小一半。

&emsp;&emsp;第一次查找范围约为：

$$
n
$$

&emsp;&emsp;第二次查找范围约为：

$$
\frac{n}{2}
$$

&emsp;&emsp;第三次查找范围约为：

$$
\frac{n}{4}
$$

&emsp;&emsp;第 $k$ 次查找范围约为：

$$
\frac{n}{2^k}
$$

&emsp;&emsp;当查找范围缩小到 1 时，算法结束，因此有：

$$
\frac{n}{2^k} = 1
$$

&emsp;&emsp;所以：

$$
2^k = n
$$

&emsp;&emsp;进一步得到：

$$
k = \log_2 n
$$

&emsp;&emsp;因此，该算法的时间复杂度为：

$$
O(\log n)
$$

````python
def find_one_local_minimum(A):
    """
    功能：
    在数组 A 中找到一个局部最小值，并返回这个局部最小值。

    题目条件：
    1. 数组中的元素互不相同
    2. A[0] > A[1]
    3. A[n - 2] < A[n - 1]

    参数：
    A：整数数组

    返回值：
    一个局部最小值
    """

    n = len(A)

    # 局部最小值需要同时有左邻居和右邻居，
    # 因此数组长度至少为 3。
    if n < 3:
        raise ValueError("数组长度至少为 3")

    # Python 数组下标从 0 开始。
    # 第一个元素下标是 0，最后一个元素下标是 n - 1。
    # 局部最小值不能是第一个元素或最后一个元素，
    # 所以搜索范围是下标 1 到 n - 2。
    left = 1
    right = n - 2

    # 使用二分查找思想，不断缩小查找范围。
    while left <= right:
        mid = (left + right) // 2

        # 情况一：
        # 如果 A[mid] 同时小于左右相邻元素，
        # 说明 A[mid] 就是局部最小值。
        if A[mid] < A[mid - 1] and A[mid] < A[mid + 1]:
            return A[mid]

        # 情况二：
        # 如果左边的相邻元素比 A[mid] 更小，
        # 说明左半部分存在局部最小值，
        # 因此继续在左半部分查找。
        elif A[mid] > A[mid - 1]:
            right = mid - 1

        # 情况三：
        # 否则说明右边的相邻元素比 A[mid] 更小，
        # 因此继续在右半部分查找。
        else:
            left = mid + 1


# 测试示例
A = [9, 3, 7, 2, 1, 4, 5]

result = find_one_local_minimum(A)

print("数组为：", A)
print("找到的一个局部最小值为：", result)
````

## 问题 5

1. 将n个数先进行排序, 然后按顺序输出最小的k个数。要求：选择合适的排序算法实现上述操作, 计算算法时间复杂度, 并使用python编程实现。
2. 建立一个包含这n个数的堆（heap）, 并且调用 k 次Extract-min 按顺序输出最小的k个数。使用往空堆中不断插入元素的方法建立堆, 分析这种方法建堆的时间复杂度, 并使用python编程实现
3. \* 假设数组中包含的数据总数目超过了计算机的存储能力, 请设计一个算法, 找到这堆数据的前k小的数值, 计算时间复杂度, 并使用python实现该算法, 假设计算机一定能存储k个数据。

e.g.：
---
数组arr=[5,4,3,2,6,1,88,33,22,107] 的前3个最小数据为：1, 2, 3

idea:

**5.1**

**排序法算法分析**

&emsp;&emsp;*核心思想*：先将数组中的所有元素按照从小到大的顺序进行排序。排序完成后，数组最前面的 $k$ 个元素就是整个数组中最小的 $k$ 个数。然后直接截取数组前 $k$ 个元素即可。

&emsp;&emsp;这里可以选择归并排序。

&emsp;&emsp;归并排序是一种基于比较的排序算法，它的主要思想是：先把数组不断分成左右两半；分到只剩一个元素时，一个元素本身就是有序的；再把两个有序数组合并成一个更大的有序数组；最终得到整个从小到大排列的数组。<br>

**时间复杂度分析**

&emsp;&emsp;归并排序会不断把数组分成左右两半。

&emsp;&emsp;数组规模的变化大致为：

$$
n,\frac{n}{2},\frac{n}{4},\frac{n}{8},\cdots
$$

&emsp;&emsp;因此，数组被不断二分，直到每个子数组只剩下 1 个元素。

&emsp;&emsp;假设一共分了 $h$ 层，则有：

$$
\frac{n}{2^h}=1
$$

&emsp;&emsp;整理得：

$$
n=2^h
$$

&emsp;&emsp;两边取对数：

$$
h=\log_2 n
$$

&emsp;&emsp;所以归并排序一共有大约：

$$
\log n
$$

&emsp;&emsp;层。

&emsp;&emsp;在每一层中，合并所有子数组时，都需要把这一层的全部元素处理一遍。

&emsp;&emsp;因此，每一层的合并代价是：

$$
O(n)
$$

&emsp;&emsp;一共有：

$$
O(\log n)
$$

&emsp;&emsp;层，所以归并排序的总时间复杂度为：

$$
O(n\log n)
$$

&emsp;&emsp;排序完成后，还需要取出前 $k$ 个元素，这一步的时间复杂度为：

$$
O(k)
$$

&emsp;&emsp;所以第一问的总时间复杂度为：

$$
O(n\log n+k)
$$

&emsp;&emsp;由于 $k$ 不会超过 $n$，并且在渐近分析中主要关注增长最快的部分，所以：

$$
O(n\log n+k)=O(n\log n)
$$

&emsp;&emsp;因此，排序法的时间复杂度为：

$$
O(n\log n)
$$
（本题的代码分析部分有借助大模型）

````python
# add your code here
def merge_sort(arr):
    """
    使用归并排序将数组从小到大排序。

    参数：
        arr: 原始数组

    返回：
        一个从小到大排序后的新数组
    """

    # 如果数组长度小于等于 1，说明数组已经有序，直接返回
    if len(arr) <= 1:
        return arr

    # 找到数组中间位置
    mid = len(arr) // 2

    # 将数组分成左半部分和右半部分
    left_part = arr[:mid]
    right_part = arr[mid:]

    # 分别对左半部分和右半部分进行归并排序
    left_sorted = merge_sort(left_part)
    right_sorted = merge_sort(right_part)

    # 将两个已经有序的数组合并成一个有序数组
    return merge(left_sorted, right_sorted)


def merge(left, right):
    """
    合并两个从小到大排序的数组。

    参数：
        left: 已经有序的左数组
        right: 已经有序的右数组

    返回：
        合并后的有序数组
    """

    result = []

    # i 指向 left 数组当前比较的位置
    # j 指向 right 数组当前比较的位置
    i = 0
    j = 0

    # 当两个数组都还有元素时，逐个比较
    while i < len(left) and j < len(right):

        # 如果 left[i] 更小，就把 left[i] 放入结果数组
        if left[i] < right[j]:
            result.append(left[i])
            i += 1

        # 否则把 right[j] 放入结果数组
        else:
            result.append(right[j])
            j += 1

    # 如果 left 数组还有剩余元素，直接加入结果数组
    while i < len(left):
        result.append(left[i])
        i += 1

    # 如果 right 数组还有剩余元素，直接加入结果数组
    while j < len(right):
        result.append(right[j])
        j += 1

    return result


def get_k_smallest_by_sort(arr, k):
    """
    先将数组从小到大排序，再返回最小的 k 个数。

    参数：
        arr: 原始数组
        k: 需要找出的最小元素个数

    返回：
        最小的 k 个数，按从小到大顺序输出
    """

    # 判断 k 是否合法
    if k <= 0:
        return []

    if k > len(arr):
        raise ValueError("k 不能大于数组长度")

    # 第一步：从小到大排序
    sorted_arr = merge_sort(arr)

    # 第二步：取前 k 个元素
    return sorted_arr[:k]


# 测试
arr = [5, 4, 3, 2, 6, 1, 88, 33, 22, 107]
k = 3

answer = get_k_smallest_by_sort(arr, k)

print("排序法得到的最小 k 个数为：", answer)
````

**最小堆算法分析思路**

&emsp;&emsp;*题目中特别强调*：使用往空堆中不断插入元素的方法建立堆。所以这里不能直接使用一次性 heapify 建堆，而应该先创建一个空堆，然后把数组中的 $n$ 个元素一个一个插入到堆中。而由于我们要找的是最小的 $k$ 个数，所以应该使用最小堆。

&emsp;&emsp;*为什么使用最小堆？*

&emsp;&emsp;最小堆的一个性质是：堆顶元素一定是当前堆中最小的元素。

&emsp;&emsp;也就是说，只要我们把所有数组元素都放入最小堆中，那么堆顶就是整个数组中的最小值。

&emsp;&emsp;当我们执行一次 Extract-min 时，就会取出当前堆中的最小值。取出之后，堆会重新调整，使新的最小值再次来到堆顶。

&emsp;&emsp;所以，只要执行 $k$ 次 Extract-min，就可以按从小到大的顺序得到最小的 $k$ 个数。

&emsp;&emsp;*算法的核心思想是*：先把所有元素逐个插入一个最小堆中，使堆顶始终保存当前最小值；然后连续执行 $k$ 次 Extract-min，每次取出当前堆中最小的元素；最终得到的 $k$ 个元素就是数组中最小的 $k$ 个数，并且输出顺序是从小到大的。

**时间复杂度分析**

&emsp;&emsp;*1. 建堆时间复杂度*

&emsp;&emsp;每插入一个元素，最坏情况下，这个元素可能需要从堆底一路向上调整到堆顶。堆是一棵接近平衡的二叉树。

&emsp;&emsp;如果堆中有 $n$ 个元素，那么堆的高度大约是：

$$
\log n
$$

&emsp;&emsp;所以，一次插入操作的最坏时间复杂度是：

$$
O(\log n)
$$

&emsp;&emsp;一共有 $n$ 个元素需要插入，因此建堆的时间复杂度为：

$$
O(n\log n)
$$

&emsp;&emsp;*2. Extract-min 时间复杂度*

&emsp;&emsp;建好最小堆之后，需要执行 $k$ 次 Extract-min 操作。每一次 Extract-min 会取出当前堆顶的最小元素。

&emsp;&emsp;取出堆顶后，需要把最后一个元素放到堆顶，然后向下调整，使堆重新满足最小堆的性质。

&emsp;&emsp;一次向下调整最多经过堆的高度，也就是：

$$
O(\log n)
$$

&emsp;&emsp;所以，一次 Extract-min 的时间复杂度是：

$$
O(\log n)
$$

&emsp;&emsp;执行 $k$ 次 Extract-min，因此这部分的时间复杂度为：

$$
O(k\log n)
$$

&emsp;&emsp;*3. 总时间复杂度*

&emsp;&emsp;建堆需要：

$$
O(n\log n)
$$

&emsp;&emsp;执行 $k$ 次 Extract-min 需要：

$$
O(k\log n)
$$

&emsp;&emsp;所以总时间复杂度为：

$$
O(n\log n+k\log n)
$$

&emsp;&emsp;可以合并为：

$$
O((n+k)\log n)
$$

&emsp;&emsp;因此，堆法的时间复杂度为：

$$
O((n+k)\log n)
$$

（本题在最小堆的代码编辑方面有参考大模型）

````python
class MinHeap:
    """
    自己实现一个最小堆。
    """

    def __init__(self):
        """
        初始化一个空堆。
        """
        self.heap = []

    def insert(self, value):
        """
        向最小堆中插入一个元素。

        参数：
            value: 要插入的元素
        """

        # 先把新元素放到数组最后
        self.heap.append(value)

        # 然后向上调整，使它回到正确位置
        self._sift_up(len(self.heap) - 1)

    def _sift_up(self, index):
        """
        向上调整。

        当新插入的元素比父节点更小时，就和父节点交换。
        """

        while index > 0:
            # 计算父节点下标
            parent = (index - 1) // 2

            # 如果当前节点比父节点小，就交换
            if self.heap[index] < self.heap[parent]:
                self.heap[index], self.heap[parent] = self.heap[parent], self.heap[index]

                # 交换后，继续向上检查
                index = parent

            # 如果当前节点不小于父节点，说明位置已经正确
            else:
                break

    def extract_min(self):
        """
        取出并删除堆中的最小元素。

        返回：
            当前堆中的最小值
        """

        # 如果堆为空，不能取最小值
        if len(self.heap) == 0:
            raise IndexError("堆为空，无法执行 Extract-min")

        # 如果堆中只有一个元素，直接弹出
        if len(self.heap) == 1:
            return self.heap.pop()

        # 堆顶元素是最小值，先保存它
        min_value = self.heap[0]

        # 把最后一个元素放到堆顶
        self.heap[0] = self.heap.pop()

        # 然后向下调整，使堆重新满足最小堆性质
        self._sift_down(0)

        # 返回原来的最小值
        return min_value

    def _sift_down(self, index):
        """
        向下调整。

        当堆顶元素比孩子节点大时，需要和较小的孩子交换。
        """

        n = len(self.heap)

        while True:
            # 当前节点的左孩子下标
            left_child = 2 * index + 1

            # 当前节点的右孩子下标
            right_child = 2 * index + 2

            # 假设当前节点是最小的
            smallest = index

            # 如果左孩子存在，并且左孩子更小
            if left_child < n and self.heap[left_child] < self.heap[smallest]:
                smallest = left_child

            # 如果右孩子存在，并且右孩子更小
            if right_child < n and self.heap[right_child] < self.heap[smallest]:
                smallest = right_child

            # 如果最小的不是当前节点，说明需要交换
            if smallest != index:
                self.heap[index], self.heap[smallest] = self.heap[smallest], self.heap[index]

                # 交换后继续向下检查
                index = smallest

            # 如果当前节点已经比两个孩子都小，调整结束
            else:
                break


def get_k_smallest_by_heap(arr, k):
    """
    使用最小堆找出最小的 k 个数。

    建堆方式：
        从空堆开始，不断插入元素。

    参数：
        arr: 原始数组
        k: 需要找出的最小元素个数

    返回：
        最小的 k 个数，按从小到大顺序输出
    """

    if k <= 0:
        return []

    if k > len(arr):
        raise ValueError("k 不能大于数组长度")

    # 第一步：建立一个空的最小堆
    min_heap = MinHeap()

    # 第二步：把数组中的元素逐个插入最小堆
    for num in arr:
        min_heap.insert(num)

    # 第三步：执行 k 次 Extract-min
    result = []

    for _ in range(k):
        smallest = min_heap.extract_min()
        result.append(smallest)

    return result


# 测试
arr = [5, 4, 3, 2, 6, 1, 88, 33, 22, 107]
k = 3

answer = get_k_smallest_by_heap(arr, k)

print("堆法得到的最小 k 个数为：", answer)
````

## 问题 6

> 星球碰撞问题：现有n个星球，在同一条直线上运行，如数组A所示，元素的绝对值表示星球的质量，负数表示星球自右向左运动，正数表示星球自左向右运动，当两个星球相撞的时候，质量小的会消失，大的保持不变，**质量相同的两个星球碰撞后自右向左运动的星球消失，自左向右的星球保持不变**，假设所有星球的速度大小相同。
>
> $ A=[23,-8, 9, -3, -7, 9, -23, 22] $

请设计一个算法模拟星球的运行情况，输出最终的星球存续情况（输出一个数组），分析算法设计思路，计算时间复杂度，并基于python编程实现。

e.g.
---
1.  输入： A=[-3,-6,2,8, 5,-8,9,-2,1]， 输出：[-3, -6, 2, 8, 9, 1]

2. 输入：A=[23,-8, 9, -3, -7, 9, -23, 22], 输出：[23, 22]

**算法设计思路

*1. 什么情况下会发生碰撞？*

&emsp;&emsp;本题中，每个星球都在同一条直线上运动。数组中的正数表示星球从左向右运动，负数表示星球从右向左运动。

&emsp;&emsp;两个星球只有在“面对面运动”时才会发生碰撞。也就是说，只有当左边的星球向右运动，右边的星球向左运动时，它们才可能相撞。

&emsp;&emsp;用数组表示，就是前一个存活的星球为正数，当前星球为负数：

<div align="center">
stack[-1] > 0 and current < 0
</div>

&emsp;&emsp;其中，`stack[-1]` 表示当前栈顶星球，也就是当前星球左边最近的、仍然存活的星球；`current` 表示正在处理的星球。

*2. 不会发生碰撞的情况*

&emsp;&emsp;下面几种情况都不会发生碰撞。<br>

&emsp;&emsp;第一种情况：-3,5<br>

&emsp;&emsp;&emsp;&emsp;左边星球向左运动，右边星球向右运动，二者背向而行，不会相撞。<br>

&emsp;&emsp;第二种情况：3,5<br>

&emsp;&emsp;&emsp;&emsp;两个星球都向右运动，方向相同，不会相撞。<br>

&emsp;&emsp;第三种情况：-3,-5<br>
&emsp;&emsp;&emsp;&emsp;两个星球都向左运动，方向相同，也不会相撞。<br>

&emsp;&emsp;因此，本题判断碰撞的核心条件就是：
<div align="center">
stack and stack[-1] > 0 and current < 0
</div>

&emsp;&emsp;只有当这个条件成立时，才需要模拟星球碰撞。

**算法核心思想**

*1. 为什么使用栈？*

&emsp;&emsp;本题适合使用栈来模拟星球碰撞过程。

&emsp;&emsp;栈可以理解为一个“保存当前仍然存活星球的数组”。我们从左到右遍历原数组，每遇到一个新的星球，就判断它是否会和栈顶星球发生碰撞。

&emsp;&emsp;之所以只需要和栈顶星球比较，是因为当前星球如果要和左边的星球发生碰撞，一定会先遇到离它最近的那个仍然存活的星球，而这个星球正好就是栈顶元素。

*2. 栈中保存什么？*

&emsp;&emsp;栈中保存的是目前已经处理过、并且还没有被撞毁的星球。

&emsp;&emsp;例如，当前栈为：stack = [23, 9]

&emsp;&emsp;这表示质量为 `23`、方向向右的星球和质量为 `9`、方向向右的星球目前都还存活。

&emsp;&emsp;如果此时来了一个当前星球：current = -23。由于栈顶星球 `9` 向右运动，而当前星球 `-23` 向左运动，所以二者会发生碰撞。

*3. 碰撞时如何处理？*

&emsp;&emsp;发生碰撞后，需要比较两个星球的质量，也就是比较它们绝对值的大小。

&emsp;&emsp;第一种情况：栈顶星球质量小。

&emsp;&emsp;如果栈顶星球质量小于当前星球质量：abs(stack[-1]) < abs(current)

&emsp;&emsp;说明栈顶星球会消失，当前星球继续向左运动，可能还会和新的栈顶星球继续发生碰撞。因此需要执行：stack.pop()

&emsp;&emsp;第二种情况：栈顶星球质量大。

&emsp;&emsp;如果栈顶星球质量大于当前星球质量：abs(stack[-1]) > abs(current)

&emsp;&emsp;说明当前星球会消失，栈顶星球保持不变。此时当前星球不能入栈，碰撞结束。

&emsp;&emsp;第三种情况：两个星球质量相同。

&emsp;&emsp;如果两个星球质量相同：abs(stack[-1]) == abs(current)

&emsp;&emsp;根据题目规则，质量相同的两个星球碰撞后，向左运动的星球消失，向右运动的星球保持不变。

&emsp;&emsp;在当前碰撞条件下，栈顶星球是向右运动的，当前星球是向左运动的，所以当前星球消失，栈顶星球保留。

*4. 为什么需要 while 循环？*

&emsp;&emsp;因为一个质量较大的向左运动星球，可能会连续撞毁多个向右运动的小星球。

&emsp;&emsp;例如：[2, 5, 8, -10]
&emsp;&emsp;处理 `-10` 时，它会先和 `8` 碰撞，`8` 消失；然后继续和 `5` 碰撞，`5` 消失；最后继续和 `2` 碰撞，`2` 消失。

&emsp;&emsp;因此，当前星球可能不止发生一次碰撞，必须使用 `while` 循环不断和新的栈顶星球比较。

*5. 算法整体流程*

&emsp;&emsp;整个算法可以概括为：

```text
1. 创建一个空栈 stack；
2. 从左到右遍历数组中的每一个星球 current；
3. 如果栈顶星球向右，当前星球向左，则发生碰撞；
4. 碰撞时比较二者质量；
5. 如果栈顶星球质量小，则弹出栈顶，当前星球继续比较；
6. 如果栈顶星球质量大，则当前星球消失；
7. 如果质量相同，则当前向左星球消失，栈顶向右星球保留；
8. 如果当前星球没有消失，则将其压入栈中；
9. 遍历结束后，栈中剩下的星球就是最终存活的星球。
```
**算法时间复杂度分析**

&emsp;&emsp;虽然代码中有 `for` 循环和 `while` 循环，看起来像是两层循环，但实际总时间复杂度不是 $O(n^2)$。

&emsp;&emsp;原因是每个星球最多只会入栈一次，也最多只会出栈一次。

&emsp;&emsp;一个星球一旦被弹出栈，就不会再回到栈中。因此，所有星球的入栈次数最多为：

$$
n
$$

&emsp;&emsp;所有星球的出栈次数最多也为：

$$
n
$$

&emsp;&emsp;所以总操作次数与星球数量成正比，时间复杂度为：

$$
O(n)
$$
(本题在算法设计分析方面参考了大模型的思路)

idea:

````python
# add your code here
def asteroid_collision(A):
    """
    功能：
        模拟星球碰撞过程，返回最终存活的星球数组。

    参数：
        A: list[int]
           数组中的每个元素表示一个星球。
           正数表示向右运动，负数表示向左运动。
           绝对值表示星球质量。

    返回：
        stack: list[int]
           最终存活的星球数组。
    """

    # 栈用来保存当前还没有被撞毁的星球
    stack = []

    # 从左到右依次处理每一个星球
    for current in A:

        # alive 表示当前星球是否还存活
        # 一开始当前星球还没有发生碰撞，所以设为 True
        alive = True

        # 只有当下面三个条件同时满足时，才会发生碰撞：
        # 1. 栈不为空
        # 2. 栈顶星球向右运动，也就是 stack[-1] > 0
        # 3. 当前星球向左运动，也就是 current < 0
        while alive and stack and stack[-1] > 0 and current < 0:

            # 栈顶星球的质量
            top_mass = abs(stack[-1])

            # 当前星球的质量
            current_mass = abs(current)

            # 情况一：栈顶星球质量小
            # 栈顶星球消失，当前星球继续向左运动
            # 所以弹出栈顶，继续和新的栈顶比较
            if top_mass < current_mass:
                stack.pop()

            # 情况二：栈顶星球质量大
            # 当前星球消失，栈顶星球保留
            elif top_mass > current_mass:
                alive = False

            # 情况三：两个星球质量相同
            # 按照题目规则：向左运动的星球消失，向右运动的星球保持不变
            # 此时 current 是向左运动的，所以 current 消失
            else:
                alive = False

        # 如果当前星球经过所有可能碰撞后仍然存活
        # 就把它加入栈中
        if alive:
            stack.append(current)

    # 栈中剩下的星球就是最终存活的星球
    return stack


# 测试示例 1
A1 = [-3, -6, 2, 8, 5, -8, 9, -2, 1]
print("示例 1 输出：", asteroid_collision(A1))

# 测试示例 2
A2 = [23, -8, 9, -3, -7, 9, -23, 22]
print("示例 2 输出：", asteroid_collision(A2))
````

## 问题 7

> 0-1背包问题：现在有4块大理石，每个大理石对应的重量和价值使用一个元组表示, 即（重量，价值），4块大理石的重量和价值为：[(5,10), (4,40), (6,30), (3, 50)], 假设你有一辆最大承重为9的小推车，请问使用这个小推车装大理石的最大价值为多少

请设计一个算法解决该问题，分析算法设计思路，计算时间复杂度，并基于python编程实现

idea:

**算法设计思路**

*1. 对“0-1 背包问题”的理解*

&emsp;&emsp;因为每块大理石只有两种选择：选它 or 不选它

&emsp;&emsp;不能选半块，也不能重复选同一块。所以每块大理石的状态只有：0-不选；1-选

&emsp;&emsp;因此这个问题叫 **0-1 背包问题**。

*2. 本题选择用动态规划的原因*

&emsp;&emsp;本题不能简单地只选价值最高的大理石，也不能简单地只选单位重量价值最高的大理石。

&emsp;&emsp;因为某一块大理石单独看可能很划算，但它可能占用太多重量，导致其他更合适的组合无法被选择。

&emsp;&emsp;所以本题更适合使用 **动态规划**。<br>

&emsp;&emsp;动态规划的核心思想是：把大问题拆成一个个小问题，先求出小问题的最优解，再逐步得到整个问题的最优解。

*3. 状态定义*

&emsp;&emsp;定义一个二维数组：

<div align="center">
dp[i][w]
</div>

&emsp;&emsp;它表示：只考虑前 i 块大理石，在小推车最大承重为 w 的情况下，能够获得的最大价值。

&emsp;&emsp;其中，`i` 表示当前考虑到第几块大理石，`w` 表示当前允许的最大承重。

&emsp;&emsp;例如：

<div align="center">
dp[3][7]
</div>

&emsp;&emsp;表示只考虑前 3 块大理石，在最大承重为 7 的情况下，最多能获得多少价值。

*4. 状态转移思路*

&emsp;&emsp;当我们考虑第 `i` 块大理石时，有两种选择。

&emsp;&emsp;第一种情况：不选第 `i` 块大理石。

&emsp;&emsp;如果不选它，那么最大价值就等于只考虑前 `i - 1` 块大理石时的结果：

<div align="center">
dp[i - 1][w]
</div>

&emsp;&emsp;第二种情况：选择第 `i` 块大理石。

&emsp;&emsp;如果选择它，就要先保证当前承重 `w` 能放下它。假设第 `i` 块大理石的重量为 `weight`，价值为 `value`，那么选择它后的价值为：

<div align="center">
dp[i - 1][w - weight] + value
</div>

&emsp;&emsp;也就是说，先给这块大理石留出重量 `weight`，剩下的容量 `w - weight` 用来装前面的石头。

&emsp;&emsp;因此，如果当前承重可以放下第 `i` 块大理石，则状态转移公式为：

$$
dp[i][w] = \max(dp[i-1][w],\ dp[i-1][w-weight] + value)
$$

&emsp;&emsp;如果当前承重放不下第 `i` 块大理石，则只能不选它：

$$
dp[i][w] = dp[i-1][w]
$$

**算法逻辑分析**

*1. 建立 dp 表*

&emsp;&emsp;因为一共有 `n` 块大理石，小推车最大承重为 `capacity`，所以建立一个大小为：

<div align="center">
(n + 1) × (capacity + 1)
</div>

&emsp;&emsp;的二维数组。

&emsp;&emsp;多出来的一行表示“不考虑任何大理石”的情况，多出来的一列表示“承重为 0”的情况。

*2. 初始化*

&emsp;&emsp;当没有任何大理石可以选时，最大价值为 0。

&emsp;&emsp;当小推车承重为 0 时，也不能装任何大理石，所以最大价值也是 0。

*3. 逐个考虑每块大理石*

&emsp;&emsp;对于每块大理石，都依次判断在不同承重限制下，是选择它更好，还是不选择它更好。

*4. 得到最终答案*

&emsp;&emsp;最终答案就是：

<div align="center">
dp[n][capacity]
</div>

&emsp;&emsp;也就是考虑所有大理石，并且最大承重为 `9` 时，能够获得的最大价值。
   <div align="center">    
    selected_stones.reverse()

    return max_value, selected_stones
   </div>


**算法时间复杂度分析**

&emsp;&emsp;设大理石数量为 `n`，小推车最大承重为 `capacity`。

&emsp;&emsp;在动态规划中，我们需要填写一个二维表 `dp`。这个表一共有：n + 1 行;&nbsp;&nbsp;&nbsp;capacity + 1 列。

&emsp;&emsp;所以需要计算的状态数量大约是：

$$
n \times capacity
$$

&emsp;&emsp;对于每一个状态 `dp[i][w]`，只需要进行常数次操作，例如比较、加法和取最大值。

&emsp;&emsp;因此，**总时间复杂度**为：

$$
O(n \times capacity)
$$

&emsp;&emsp;在本题中：n = 4，capacity = 9

&emsp;&emsp;所以状态数量大约为：

$$
4 \times 9 = 36
$$

&emsp;&emsp;也就是说，本题规模很小，计算量并不大。<br>
（本题在0-1背包问题的理解以及代码分析上面参考了大模型）

````python
# add your code here
def knapsack_01(stones, capacity):
    """
    功能：
        使用动态规划解决 0-1 背包问题，
        计算在不超过小推车最大承重的情况下，可以获得的最大价值。

    参数：
        stones: list[tuple[int, int]]
            每个元素是一个元组，形式为 (重量, 价值)

        capacity: int
            小推车的最大承重

    返回：
        max_value: int
            能获得的最大价值

        selected_stones: list[tuple[int, int]]
            被选择的大理石列表
    """

    # n 表示大理石的数量
    n = len(stones)

    # 创建 dp 表
    # dp[i][w] 表示：只考虑前 i 块大理石，在承重为 w 时的最大价值
    dp = [[0 for _ in range(capacity + 1)] for _ in range(n + 1)]

    # 开始填表
    # i 从 1 开始，表示正在考虑第 i 块大理石
    for i in range(1, n + 1):

        # 取出第 i 块大理石的重量和价值
        # 因为 stones 的下标从 0 开始，所以第 i 块对应 stones[i-1]
        weight, value = stones[i - 1]

        # w 表示当前小推车允许的最大承重
        for w in range(1, capacity + 1):

            # 情况一：当前承重 w 放不下这块大理石
            # 那么只能不选它
            if weight > w:
                dp[i][w] = dp[i - 1][w]

            # 情况二：当前承重 w 可以放下这块大理石
            else:
                # 不选第 i 块大理石时的最大价值
                not_choose = dp[i - 1][w]

                # 选择第 i 块大理石时的最大价值
                choose = dp[i - 1][w - weight] + value

                # 两种情况取最大值
                dp[i][w] = max(not_choose, choose)

    # dp[n][capacity] 就是最终最大价值
    max_value = dp[n][capacity]

    # 下面用于反推出具体选择了哪些大理石
    selected_stones = []

    # 从最后一个状态开始往回找
    w = capacity

    for i in range(n, 0, -1):

        # 如果 dp[i][w] != dp[i-1][w]
        # 说明第 i 块大理石被选中了
        if dp[i][w] != dp[i - 1][w]:

            weight, value = stones[i - 1]

            # 将这块大理石加入选择结果
            selected_stones.append((weight, value))

            # 减去这块大理石占用的重量
            w -= weight

    # 因为我们是从后往前找的，所以最后反转一下顺序
    selected_stones.reverse()

    return max_value, selected_stones


# 测试数据
stones = [(5, 10), (4, 40), (6, 30), (3, 50)]
capacity = 9

max_value, selected_stones = knapsack_01(stones, capacity)

print("最大价值为：", max_value)
print("选择的大理石为：", selected_stones)
````

输入：root = [1,3,null,null,2]， 输出为：[3,1,null,null,2]， 

解释：3 不能是 1 的左孩子，因为 3 > 1 。交换 1 和 3 使二叉搜索树有效。

 <div style="text-align:center;width:200px;"> <img alt="" src="./fig/hw3q9-1.jpg"> </div>
输入： root=[3,1,4,null,null,2]， 输出为：[2,1,4,null,null,3]

解释：2 不能在 3 的右子树中，因为 2 < 3 。交换 2 和 3 使二叉搜索树有效。

idea:

**题目分析**

&emsp;&emsp;本题给定一棵二叉搜索树 `root`，但是其中恰好有两个节点的值被错误地交换了。现在要求在不改变树结构的情况下，恢复这棵二叉搜索树。

&emsp;&emsp;二叉搜索树有一个非常重要的性质：

$$
左子树所有节点值 < 根节点值 < 右子树所有节点值
$$

&emsp;&emsp;如果对一棵二叉搜索树进行中序遍历，那么得到的节点值序列一定是从小到大排列的。

&emsp;&emsp;例如，一棵正常的二叉搜索树中序遍历结果可能是：

$$
1,\ 2,\ 3,\ 4,\ 5
$$

&emsp;&emsp;但是如果其中两个节点的值被交换了，中序遍历结果就会出现顺序错误。例如：

$$
1,\ 4,\ 3,\ 2,\ 5
$$

&emsp;&emsp;这里可以看到，`4 > 3`，`3 > 2`，说明中序遍历序列中出现了不符合升序的位置。

&emsp;&emsp;因此，本题的关键就是：通过中序遍历找到这两个被错误交换的节点，然后交换它们的值，而不是改变节点之间的连接关系。

**算法设计思路**

*1. 为什么使用中序遍历？*

&emsp;&emsp;因为二叉搜索树的中序遍历结果应该是一个严格递增序列。

&emsp;&emsp;也就是说，对于正常的二叉搜索树，中序遍历时前一个节点值应该小于当前节点值：

$$
prev.val < current.val
$$

&emsp;&emsp;如果在遍历过程中发现：

$$
prev.val > current.val
$$

&emsp;&emsp;就说明这两个位置出现了顺序错误。

*2. 被交换节点会造成什么现象？*

&emsp;&emsp;如果两个错误节点在中序序列中相邻，那么只会出现一次逆序。

&emsp;&emsp;例如：

$$
1,\ 3,\ 2,\ 4
$$

&emsp;&emsp;这里 `3` 和 `2` 被交换了，只出现一次错误：

$$
3 > 2
$$

&emsp;&emsp;此时第一个错误节点是 `3`，第二个错误节点是 `2`。

&emsp;&emsp;如果两个错误节点在中序序列中不相邻，那么会出现两次逆序。

&emsp;&emsp;例如：

$$
1,\ 4,\ 3,\ 2,\ 5
$$

&emsp;&emsp;第一次错误是：

$$
4 > 3
$$

&emsp;&emsp;第二次错误是：

$$
3 > 2
$$

&emsp;&emsp;真正被交换的两个节点是第一次逆序中的较大值 `4`，以及第二次逆序中的较小值 `2`。

*3. 如何记录错误节点？*

&emsp;&emsp;在中序遍历过程中，用 `prev` 记录上一个访问过的节点，用 `current` 表示当前访问到的节点。

&emsp;&emsp;当发现：

$$
prev.val > current.val
$$

&emsp;&emsp;说明出现了逆序。

&emsp;&emsp;第一次发现逆序时：

&emsp;&emsp;把 `prev` 记录为第一个错误节点 `first`，因为它本来应该更靠后，却出现在了前面。

&emsp;&emsp;同时把 `current` 记录为第二个错误节点 `second`。如果之后又发现第二次逆序，就更新 `second` 为当前节点 `current`。

&emsp;&emsp;最后只需要交换：

$$
first.val
$$

&emsp;&emsp;和：

$$
second.val
$$

&emsp;&emsp;就可以恢复这棵二叉搜索树。

*4. 为什么只交换值，不改变结构？*

&emsp;&emsp;题目要求在不改变二叉树结构的情况下恢复二叉搜索树。所以我们不能移动节点，也不能重新连接左右孩子。

&emsp;&emsp;正确做法是只交换两个错误节点的值：

$$
first.val,\ second.val = second.val,\ first.val
$$

&emsp;&emsp;这样树的形状完全不变，只是把错误的值改回正确位置。

**详细执行步骤**

*1. 进行中序遍历*

&emsp;&emsp;中序遍历顺序为：

$$
左子树 \rightarrow 根节点 \rightarrow 右子树
$$

&emsp;&emsp;对于二叉搜索树来说，中序遍历应该得到升序序列。

*2. 用 prev 记录前一个节点*

&emsp;&emsp;每访问一个节点 `current`，就和前一个节点 `prev` 比较。

&emsp;&emsp;如果：

$$
prev.val > current.val
$$

&emsp;&emsp;说明这里出现了顺序错误。

*3. 找到两个错误节点*

&emsp;&emsp;第一次发现逆序时，记录：

$$
first = prev
$$

$$
second = current
$$

&emsp;&emsp;如果后面再次发现逆序，就更新：

$$
second = current
$$

&emsp;&emsp;这样可以同时处理错误节点相邻和不相邻两种情况。

*4. 交换两个错误节点的值*

&emsp;&emsp;中序遍历结束后，`first` 和 `second` 就是被错误交换的两个节点。

&emsp;&emsp;交换它们的值：

$$
first.val,\ second.val = second.val,\ first.val
$$

&emsp;&emsp;恢复完成。

**算法时间复杂度分析**

*1. 用例子理解每个节点只会被访问一次*

&emsp;&emsp;例如下面这棵树：

<div align="center">
<pre style="display: inline-block; text-align: left;">
      3
     / \
    1   4
       /
      2
</pre>
</div>

&emsp;&emsp;对它进行中序遍历时，访问顺序是：1 → 3 → 2 → 4。在这个过程中，算法会依次访问节点 `1`、`3`、`2`、`4`。

&emsp;&emsp;每个节点只会被访问一次，不会重复访问。

&emsp;&emsp;也就是说，虽然算法需要比较相邻节点的大小关系，但是这些比较都是在一次中序遍历过程中完成的，并不会对同一个节点反复遍历。

*2. 每个节点上做了哪些操作？*

&emsp;&emsp;当中序遍历访问到一个节点 `current` 时，算法主要做以下几件事：

&emsp;&emsp;第一，判断前一个节点 `prev` 是否存在。

&emsp;&emsp;第二，比较 `prev.val` 和 `current.val` 的大小。

&emsp;&emsp;第三，如果发现逆序，就更新 `first` 和 `second` 这两个错误节点。

&emsp;&emsp;第四，将 `prev` 更新为当前节点。

&emsp;&emsp;这些操作都只是简单的比较和赋值操作，执行时间不会随着树的规模变大而单独增加。

&emsp;&emsp;所以，对于每一个节点来说，处理它所需要的时间可以看成：O(1)

*3. 正式推导时间复杂度*

&emsp;&emsp;设二叉搜索树中一共有 `n` 个节点。

&emsp;&emsp;中序遍历会访问整棵树中的每一个节点。因为每个节点只访问一次，所以访问节点的总次数是：n

&emsp;&emsp;每个节点被访问时，只进行常数次操作，因此每个节点的处理时间是：O(1)

&emsp;&emsp;所以总时间可以表示为：n × O(1)。因此，算法的**时间复杂度**为：O(n)

（本题中的算法不是先遍历一遍，再重新遍历一遍找错误，而是在访问每个节点时直接和前一个节点比较。所以整个过程只需要一次完整的中序遍历。因此时间复杂度不是：O(n²)，而是：&nbsp;O(n) ）

*4. 交换两个节点的值会不会影响时间复杂度？*

&emsp;&emsp;中序遍历结束后，算法已经找到了两个被错误交换的节点 `first` 和 `second`。

&emsp;&emsp;最后只需要交换它们的值：first.val, second.val = second.val, first.val

&emsp;&emsp;交换两个变量的值只需要常数时间，所以这一步的时间复杂度是：O(1)

&emsp;&emsp;它不会改变整个算法的总时间复杂度。

*5. 总结*

&emsp;&emsp;本算法通过一次中序遍历找到两个被错误交换的节点。遍历过程中，每个节点只会被访问一次，每次访问只进行常数次比较和赋值操作，因此总时间复杂度为:&nbsp;O(n)

&emsp;&emsp;其中，`n` 表示二叉搜索树中的节点总数。<br>

````python
# add your code here
from collections import deque


class TreeNode:
    """
    二叉树节点类。
    每个节点包含：
    val: 节点值
    left: 左孩子
    right: 右孩子
    """

    def __init__(self, val=0, left=None, right=None):
        self.val = val
        self.left = left
        self.right = right


def recover_bst(root):
    """
    功能：
        恢复一棵有两个节点值被错误交换的二叉搜索树。

    参数：
        root: TreeNode
            二叉搜索树的根节点。

    返回：
        root: TreeNode
            恢复后的二叉搜索树根节点。
    """

    # first 用来记录第一个错误节点
    first = None

    # second 用来记录第二个错误节点
    second = None

    # prev 用来记录中序遍历过程中当前节点的前一个节点
    prev = None

    def inorder(node):
        """
        中序遍历函数。
        遍历顺序：左子树 -> 当前节点 -> 右子树。
        """
        nonlocal first, second, prev

        # 如果当前节点为空，直接返回
        if node is None:
            return

        # 先遍历左子树
        inorder(node.left)

        # 访问当前节点时，判断是否出现逆序
        # 正常二叉搜索树的中序遍历应该是递增的
        # 如果 prev.val > node.val，说明出现错误
        if prev is not None and prev.val > node.val:

            # 如果 first 还没有记录，说明这是第一次发现逆序
            # 第一个错误节点应该是 prev
            if first is None:
                first = prev

            # 不管是第一次还是第二次发现逆序
            # 第二个错误节点都应该更新为当前 node
            second = node

        # 更新 prev，让当前节点成为下一个节点的前一个节点
        prev = node

        # 最后遍历右子树
        inorder(node.right)

    # 对整棵树进行中序遍历，找出两个错误节点
    inorder(root)

    # 如果找到了两个错误节点，就交换它们的值
    if first is not None and second is not None:
        first.val, second.val = second.val, first.val

    return root


def build_tree(values):
    """
    根据层序数组构建二叉树。
    例如：
        [1, 3, None, None, 2]
    表示：
        根节点 1
        左孩子 3
        右孩子为空
        3 的左孩子为空
        3 的右孩子为 2
    """

    if not values:
        return None

    # 创建根节点
    root = TreeNode(values[0])

    # 使用队列辅助构建二叉树
    queue = deque([root])

    index = 1

    while queue and index < len(values):
        node = queue.popleft()

        # 构建左孩子
        if index < len(values) and values[index] is not None:
            node.left = TreeNode(values[index])
            queue.append(node.left)

        index += 1

        # 构建右孩子
        if index < len(values) and values[index] is not None:
            node.right = TreeNode(values[index])
            queue.append(node.right)

        index += 1

    return root


def level_order(root):
    """
    将二叉树按照层序遍历转换成数组。
    方便查看恢复后的树结构。
    """

    if root is None:
        return []

    result = []
    queue = deque([root])

    while queue:
        node = queue.popleft()

        if node is None:
            result.append(None)
        else:
            result.append(node.val)
            queue.append(node.left)
            queue.append(node.right)

    # 删除末尾多余的 None
    while result and result[-1] is None:
        result.pop()

    return result


# 测试示例 1
root1 = build_tree([1, 3, None, None, 2])
recover_bst(root1)
print("示例 1 恢复后：", level_order(root1))

# 测试示例 2
root2 = build_tree([3, 1, 4, None, None, 2])
recover_bst(root2)
print("示例 2 恢复后：", level_order(root2))
````

**算法时间复杂度分析**

*1. 先看算法到底做了什么*

&emsp;&emsp;本题使用的是深度优先搜索，也就是 DFS。算法会从根节点出发，沿着树不断往下走，直到到达叶子节点。

&emsp;&emsp;在遍历过程中，每到一个节点，只做一件很简单的事：把当前节点的数字接到已经形成的路径数字后面。

&emsp;&emsp;更新公式为：current_sum = current_sum × 10 + node.val

&emsp;&emsp;这个操作只包含一次乘法、一次加法，所以对每个节点来说，处理时间是常数级别的。

*2. 用例子理解每个节点只会被访问一次*

&emsp;&emsp;例如下面这棵树：

<div align="center">
<pre style="display: inline-block; text-align: left;">
        4
       / \
      9   0
     / \
    5   1
</pre>
</div>

&emsp;&emsp;它的根到叶子节点路径有三条：

<div align="center">
4 → 9 → 5
</div>

<div align="center">
4 → 9 → 1
</div>

<div align="center">
4 → 0
</div>

&emsp;&emsp;但是 DFS 并不是把每条路径都从根节点重新走一遍。它的实际访问过程可以理解为：4 → 9 → 5

&emsp;&emsp;到达叶子节点 `5` 后返回，再继续访问：1;&nbsp;&nbsp;然后再返回，继续访问：0

&emsp;&emsp;所以整棵树中的节点：4, 9, 0, 5, 1;&nbsp;&nbsp;每个节点都只会被访问一次。

&emsp;&emsp;也就是说，虽然最终有多条路径，但是算法不会对公共部分反复计算。例如路径 `4 → 9 → 5` 和 `4 → 9 → 1` 都经过 `4 → 9`，DFS 访问到 `9` 时只处理一次，并不会为了每条路径都重新从 `4` 开始计算。

*3. 为什么不是 O(路径数 × 路径长度)？*

&emsp;&emsp;如果我们采用一种比较笨的方法，先把所有路径都保存下来，再逐条路径拼接数字，那么可能会涉及“路径数 × 路径长度”的计算。

&emsp;&emsp;但是本题的 DFS 方法不是这样做的。它是在递归向下走的过程中，直接把当前路径数字传下去。例如：

<div align="center">
4 → 9 → 5
</div>

&emsp;&emsp;它的计算过程是：

<div align="center">
0 × 10 + 4 = 4
</div>

<div align="center">
4 × 10 + 9 = 49
</div>

<div align="center">
49 × 10 + 5 = 495
</div>

&emsp;&emsp;到了叶子节点后，直接得到完整数字 `495`，不需要再重新遍历这条路径。

&emsp;&emsp;因此，这个算法的总计算量主要取决于树中一共有多少个节点，而不是路径有多少条。

*4. 正式推导时间复杂度*

&emsp;&emsp;设二叉树中一共有 `n` 个节点。DFS 会访问整棵树中的每一个节点。

&emsp;&emsp;对于每个节点，算法只做常数次操作，例如：

&emsp;&emsp;第一，判断当前节点是否为空；

&emsp;&emsp;第二，计算新的路径数字；

&emsp;&emsp;第三，判断当前节点是否是叶子节点；

&emsp;&emsp;第四，递归访问左子树和右子树。

&emsp;&emsp;这些操作的时间不会随着 `n` 的增大而单独变大，所以每个节点的处理时间可以看作：O(1)

&emsp;&emsp;一共有 `n` 个节点，每个节点处理一次，所以总时间为：n × O(1)

&emsp;&emsp;因此，算法的**时间复杂度**为：O(n)

*5. 总结*

&emsp;&emsp;本算法虽然会计算多条根到叶子节点的路径，但它在 DFS 过程中不会重复遍历节点。每个节点最多被访问一次，每次访问只进行常数次计算，所以总时间复杂度是：O(n)

&emsp;&emsp;其中，`n` 表示二叉树中的节点总数。<br>

````python
# add your code here
from collections import deque


class TreeNode:
    """
    二叉树节点类。
    每个节点包含：
    val: 节点值
    left: 左孩子
    right: 右孩子
    """

    def __init__(self, val=0, left=None, right=None):
        self.val = val
        self.left = left
        self.right = right


def sum_numbers(root):
    """
    功能：
        计算二叉树中所有从根节点到叶子节点路径所表示数字的总和。

    参数：
        root: TreeNode
            二叉树根节点。

    返回：
        int
            所有根到叶子路径形成数字的总和。
    """

    def dfs(node, current_sum):
        """
        深度优先搜索函数。

        参数：
            node: 当前访问到的节点
            current_sum: 从根节点到当前节点之前，已经形成的数字

        返回：
            当前子树中所有根到叶子路径形成数字的总和
        """

        # 如果当前节点为空，说明这条路不存在，返回 0
        if node is None:
            return 0

        # 将当前节点的值拼接到路径数字后面
        # 例如之前是 49，当前节点是 5，则变成 495
        current_sum = current_sum * 10 + node.val

        # 如果当前节点没有左孩子，也没有右孩子
        # 说明当前节点是叶子节点
        # 此时 current_sum 就是一条完整路径形成的数字
        if node.left is None and node.right is None:
            return current_sum

        # 如果当前节点不是叶子节点
        # 就继续递归计算左子树和右子树的路径数字之和
        left_sum = dfs(node.left, current_sum)
        right_sum = dfs(node.right, current_sum)

        # 当前节点的所有路径和 = 左子树路径和 + 右子树路径和
        return left_sum + right_sum

    # 从根节点开始，初始路径数字为 0
    return dfs(root, 0)


def build_tree(values):
    """
    根据层序数组构建二叉树。

    例如：
        [4, 9, 0, 5, 1]

    表示：
            4
           / \
          9   0
         / \
        5   1
    """

    # 如果数组为空，返回空树
    if not values:
        return None

    # 创建根节点
    root = TreeNode(values[0])

    # 使用队列辅助构建二叉树
    queue = deque([root])

    # index 用来遍历数组
    index = 1

    while queue and index < len(values):
        node = queue.popleft()

        # 构建左孩子
        if index < len(values) and values[index] is not None:
            node.left = TreeNode(values[index])
            queue.append(node.left)

        index += 1

        # 构建右孩子
        if index < len(values) and values[index] is not None:
            node.right = TreeNode(values[index])
            queue.append(node.right)

        index += 1

    return root


# 测试示例 1
root1 = build_tree([1, 2, 3])
print("示例 1 输出：", sum_numbers(root1))

# 测试示例 2
root2 = build_tree([4, 9, 0, 5, 1])
print("示例 2 输出：", sum_numbers(root2))
````
